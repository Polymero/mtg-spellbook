```python
import json
import os
```

```python
pathWindows = "C:\\Users\\Nika\\Downloads\\"
pathLinux = '/home/nika/Downloads/'
pathBase = pathWindows if os.name == 'nt' else pathLinux
```

```python
SFC = json.load(open(pathBase + 'default-cards-20260123101154.json', 'rb'))
MKM = json.load(open(pathBase + 'products_singles_1.json', 'rb'))['products']
```

```python
print(len(SFC))
SFCf = [i for i in SFC if 'paper' in i['games']]
print(len(SFCf))
desiredLayouts = ["normal", "split", "flip", "transform", "modal_dfc", "meld", "leveler", "class", "case", "saga", "adventure", "mutate", "prototype", "battle", "reversible_card"]
SFCf = [i for i in SFCf if i['layout'] in desiredLayouts]
print(len(SFCf))
SFCf = [i for i in SFCf if ('promo_types' not in i) or ('thick' not in i['promo_types'])]
print(len(SFCf))
```

```python
SFCf_names = set(i['name'] for i in SFCf)
```

```python
MKMf = [i for i in MKM if i['name'] in SFCf_names]
```

```python
idExpansionSet = set()
for i in MKMf:
    idExpansionSet.add(i['idExpansion'])
```

```python
MKMf_name_by_idExpansion = { i : set() for i in idExpansionSet }
for i in MKMf:
    MKMf_name_by_idExpansion[i['idExpansion']].add(i['name'])
```

```python
len(idExpansionSet)
```

```python
SetCodeSet = set()
for i in SFCf:
    SetCodeSet.add(i['set'])
```

```python
SFCf_name_by_set = { i : set() for i in SetCodeSet }
for i in SFCf:
    SFCf_name_by_set[i['set']].add(i['name'])
```

```python
len(SetCodeSet)
```

```python
idList = list(idExpansionSet)
```

```python
setList = list(SetCodeSet)
```

```python
setCode_to_idExpansion = []
for setCode in SetCodeSet:
    target = SFCf_name_by_set[setCode]
    hitRatio = [len(MKMf_name_by_idExpansion[i] & target) / len(target) for i in idList]
    for ratio in hitRatio:
        if ratio > 0.8:
            setCode_to_idExpansion.append([setCode, idList[hitRatio.index(max(hitRatio))]])
```

```python
foundCodes, foundIds = [set(i) for i in zip(*setCode_to_idExpansion)]
print(len(foundCodes), len(foundIds))
```

```python
for idExpansion in idExpansionSet:
    target = MKMf_name_by_idExpansion[idExpansion]
    hitRatio = [len(SFCf_name_by_set[i] & target) / len(target) for i in setList]
    for ratio in hitRatio:
        if ratio > 0.8:
            setCode_to_idExpansion.append([setList[hitRatio.index(max(hitRatio))], idExpansion])
```

```python
foundCodes, foundIds = [set(i) for i in zip(*setCode_to_idExpansion)]
print(len(foundCodes), len(foundIds))
```

```python
lostCodes = SetCodeSet.difference(foundCodes)
lostIds = idExpansionSet.difference(foundIds)
print(len(lostCodes), len(lostIds))
```

```python
Code_to_Id = { i : set() for i in foundCodes }
for i in setCode_to_idExpansion:
    Code_to_Id[i[0]].add(i[1])
```

```python
Id_to_Code = { i : set() for i in foundIds }
for i in setCode_to_idExpansion:
    Id_to_Code[i[1]].add(i[0])
```
